```python
from typing_extensions import ItemsView
nomes= ["Luffy", "Sosuke", "Becka"]
idades= [1.8, 1.8, 5]
espécies= ["gato", "porquinho da india", "cachorro"]

#Verificando se o item existe na lista
print("Sosuke" in nomes)

#max,min,media
print(max(nomes))
print(min(idades))

#edita um Item
nomes[0] = "Ruffus"
print(nomes)

#remover itens
#1
nomes.remove("Ruffus")
print(nomes)
#2
nomes.pop(-1)
print(nomes)

#Adicionar itens + posição específica # O append() é usado para adiconar de forma aleatória
nomes.insert(0,"luffy")
nomes.insert(2, "Becka")
print(nomes)

#para adicionar uma lista em outra se usa: nomes.extend(idades) por ex

# ordenar uma lista
nomes.sort()
print(nomes)

nomes.sort(reverse=True)
print(nomes)

```
