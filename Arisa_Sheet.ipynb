```python
Name= "Arisa"
Age= 21
heigh = 1.64
level = 17
life = 300
mana = 300
player = "leleka"
xp = 50
potion = 0.25
dano = 35
level_up = 3
```

```python
#pontos de experiência
novo_xp = life + xp
print(novo_xp)
```

```python
#ataque que causa 35 de dado
novo_xp= novo_xp - dano
print(novo_xp)

#poção que recupera 25% da vida
recuperação = novo_xp * potion
novo_xp = novo_xp + recuperação
print(novo_xp)
```

```python
#novo nível
novo_nivel= level + level_up
print(novo_nivel)
```

```python
#dividindo a grana
money = 127
players= 5


caridade= money // players
moedas_sobrando = money % players
divisão_normal = money / players

print(caridade)
print(moedas_sobrando)
print(divisão_normal)
```

```python
#Comparações

#O personagem é maior de dezoito anos?
maior_dezoito = Age >= 18
print(maior_dezoito)

#O personagem está no level 10 ou acima?
level_dez= level >= 10
print(level_dez)

#A vida está abaixo de 50?
vital_status= life <= 50
print(vital_status)

#O personagem está ativo?
status = True
online= status
offline= not status
print(online)
print(offline)


```

```python
#operadores lógicos

pode_entrar_dungeon = level >= 10 and life >= 100
print(pode_entrar_dungeon)

pode_fugir = mana == 0 or life <= 50
print(pode_fugir)
```
