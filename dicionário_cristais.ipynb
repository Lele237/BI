```python
cristais = {}
cristais = int(input("quantos cristais você deseja adicionar?"))
print(cristais)
```

```python
for i in range(cristais):
  nome = input("nome do cristal:")
  cor = input("cor do cristal:")
  tipo = input("tipo do cristal:")
  valor = int(input("valor:"))

```

```python
cristais = {
    nome:{
        "cor" : cor,
        "tipo" : tipo,
        "valor" : valor
    }
}
```

```python
buscar_cristal = input("digite o nome do cristal:")
buscar_cristal=buscar_cristal.strip()
buscar_cristal=buscar_cristal.lower()

if buscar_cristal in cristais:
  valor = cristais[buscar_cristal]

  print("pedra encontrada")
  print(f"pedra: {buscar_cristal}, preço: R${valor}")

else:
  print("pedra não encontrada")
```
